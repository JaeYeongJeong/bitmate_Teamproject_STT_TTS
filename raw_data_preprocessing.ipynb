```python
# 녹음된 음성 데이터를 전처리하는 과정입니다.
# 녹음된 음성데이터는 비트메이트_TP01/data/dataset/wav_{speaker}/raw폴더에 있어야합니다.
# raw 폴더 내 음성데이터를 변환해서 wavs 폴더에 저장합니다.

# .m4a .wav 상관 없이 .wav 로 변환해줍니다.
# samplingrate는 24000으로 변경합니다. (finetunning시 24000으로 학습시켜야함)
# 무음 구간은 제거합니다
# 볼륨도 일정하게 조절합니다.
```

```python
# 음성데이터 폴더 경로 (필요시 수정해주세요)
speaker = 'jhc100'
base_dir = f'/content/drive/MyDrive/비트메이트_TP01/data/dataset/wav_{speaker}'
%cd {base_dir}
```

```python
import os
import librosa
import soundfile as sf
import numpy as np

input_dir = "./raw"
output_dir = "./wavs"

os.makedirs(output_dir, exist_ok=True)

TARGET_SR = 24000

for file in os.listdir(input_dir):
    if not (file.endswith(".wav") or file.endswith(".m4a")):
        continue

    input_path = os.path.join(input_dir, file)
    output_path = os.path.join(output_dir, os.path.splitext(file)[0] + ".wav")

    try:
        # 1. 로드 + mono + resample
        audio, sr = librosa.load(input_path, sr=TARGET_SR, mono=True)

        # 2. 무음 제거 (앞뒤)
        audio, _ = librosa.effects.trim(
            audio,
            top_db=30  # 값 낮을수록 더 많이 자름 (20~40 추천)
        )

        # 3. normalize (볼륨 일정하게)
        if np.max(np.abs(audio)) > 0:
            audio = audio / np.max(np.abs(audio))

        # 4. 저장
        sf.write(output_path, audio, TARGET_SR)

        print(f"완료: {file}")

    except Exception as e:
        print(f"에러: {file}, {e}")
```
