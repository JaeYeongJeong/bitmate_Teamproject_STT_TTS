```python
# qwen 모델을 불러오고 gpu 설정

import torch
from qwen_tts import Qwen3TTSModel

# MODEL_NAME = "Qwen/Qwen3-TTS-12Hz-0.6B-Base"   # 가볍게 시작
MODEL_NAME = "Qwen/Qwen3-TTS-12Hz-1.7B-Base" # 더 무겁지만 품질 기대 가능

device = "cuda:0" if torch.cuda.is_available() else "cpu"
dtype = torch.bfloat16 if torch.cuda.is_available() else torch.float32

print("device:", device)
print("dtype:", dtype)
print("loading model...")

model = Qwen3TTSModel.from_pretrained(
    MODEL_NAME,
    device_map=device,
    dtype=dtype,
)

print("model loaded")
```

```python
import soundfile as sf
from IPython.display import Audio, display

# 음성 데이터 경로
ref_audio = "/content/drive/MyDrive/비트메이트_TP01/data/dataset/wav_jhc/wavs/audio_001.wav"
# 음성 데이터 대본
ref_text  = "안녕하세요, 오늘 하루도 차분하게 시작해 보겠습니다."

# 생성
wavs, sr = model.generate_voice_clone(
    text="토끼와 거북이는 빠른 토끼가 느린 거북이를 얕보고 경주 도중 방심해 잠이 드는 사이, 꾸준히 쉬지 않고 나아간 거북이가 결국 먼저 도착해 승리한다는 이야기로, 재능이나 속도보다도 끝까지 성실하게 노력하는 태도가 더 중요하다는 교훈을 전한다.",
    language="Korean",
    ref_audio=ref_audio,
    ref_text=ref_text,
)

# 저장
sf.write("output_voice_clone.wav", wavs[0], sr)

# 재생
display(Audio("output_voice_clone.wav"))
```
